```python
train_df["group"] = train_df["day"].eq(1).cumsum()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 月の区切り（日数の累積）
month_ends = [31, 59, 90, 120, 151, 181,
              212, 243, 273, 304, 334, 365]

# day列からmonth列を作成
train_df["month"] = pd.cut(
    train_df["day"],
    bins=[0] + month_ends,
    labels=range(1, 13),
    include_lowest=True
)

# 年・日・月を除いた説明変数だけを描画
exclude_columns = ["id", "day", "month", "group"]
columns = [
    column
    for column in train_df.select_dtypes("number").columns
    if column not in exclude_columns
]

# 各列について、同じ月の年別分布を比較
for column in columns:
    fig, axes = plt.subplots(3, 4, figsize=(16, 10))
    axes = axes.ravel()

    for month in range(1, 13):
        month_data = train_df[train_df["month"] == month]

        sns.boxplot(
            data=month_data,
            x="group",
            y=column,
            ax=axes[month - 1]
        )
        axes[month - 1].set_title(f"Month {month}")
        axes[month - 1].set_xlabel("Year")
        axes[month - 1].set_ylabel(column)

    fig.suptitle(f"{column}: distribution by year for each month")
    plt.tight_layout()
    plt.show()
```
